
###  <font color="DodgerBlue">Simple Example (HF is good start point)</font>

#### 1) For most molecules, HF is a good start point. Let's setup a molecule.

```python
from trimci.molecule_setup import setup_molecule
from scipy.special import comb

#MOLECULE = "N 0 0 0; N 0 0 2.5"
MOLECULE = "Cr 0 0 0; Cr 0 0 2.5"
BASIS = 'sto-3g'
SPIN = 0

mol, mf, h1, eri = setup_molecule(MOLECULE, BASIS, SPIN)
n_orb = h1.shape[0]
n_e = mol.nelectron
n_alpha = (n_e-SPIN) // 2
n_beta = n_e - n_alpha
n_total = int(comb(n_orb, n_alpha)**2)

print(f"  Molecule string: {MOLECULE}")
print(f"  Basis set: {BASIS}")
print(f"  Number of sptail basis functions (orbitals): {n_orb}")
print(f"  Total electrons: {n_e}")
print(f"  Alpha electrons: {n_alpha}")
print(f"  Beta electrons: {n_beta}")
print(f"  Total determinants: {n_total:.2e}")
```

#### 2) Solve it with Selected-CI in PySCF

```python
from pyscf.fci import selected_ci,direct_spin1

cisolver = selected_ci.SelectedCI()
cisolver.select_cutoff = 5e-2
cisolver.verbose = 5
sci_energy, sci_vec = cisolver.kernel(h1, eri, n_orb, (n_alpha, n_beta))
sci_energy = sci_energy + mol.energy_nuc()

print(f"\n=== Selected-CI(PySCF version) Results ===")
print(f"n_dets  = {sci_vec.shape[0]**2}")
print(f"SCI Energy        = {sci_energy:.10f}")

if n_total < 1e6:
    e_fci, _ = direct_spin1.kernel(h1, eri, n_orb, (n_alpha, n_beta))
    e_fci = e_fci + mol.energy_nuc()
    error_sci = abs(e_fci - sci_energy)
    print(f"FCI Energy        = {e_fci:.10f}")
    print(f"Error (SCI)       = {error_sci:.10f}") 
```

#### 3) It takes around 1m10s and 324 dets to get energy -2064.3384791729

#### 4) Now, let's run TrimCI

```python
from trimci import run_full_calculation


config_dict = {
    "max_final_dets": 250,
    "max_rounds": 2,
    "pool_core_ratio": 10,
    "core_set_ratio": 1.1,
    "pool_build_strategy": 'normalized_uniform',
}

final_energy, final_dets, final_coeffs, details, configs = run_full_calculation(
    molecule=MOLECULE, basis=BASIS, spin=SPIN,
    #trimci_config_path=str(config_path),  # Optional: use config file
    config_dict=config_dict, # Optional: use config dict
)

print("=== TrimCI Result ===")
print(f"Molecule: {MOLECULE}, basis: {BASIS}")
print(f"Final energy (Ha): {final_energy:.8f}")
print(f"Final determinants: {len(final_dets)}")
print(f"Results dir: {details.get('results_dir', 'results')}")
```

#### 5) It takes around 2.3s and 280 dets to get energy -2064.43563478

#### Much faster, Much more accurate.

### <font color="DodgerBlue">Yet, TrimCI can also solve it from random initial determinants</font>

#### 6) What if we pretend that we don't know Hatree-Fock state? Can we find it in the space of 1.57e+18 determinants?

#### 7) Let's run TrimCI with 50 indepdent runs with 100 random initial determinants each

#### It takes a bit longer since we don't have prior knowledge.

```python
from trimci import run_full_calculation

config_dict = {
    "initial_dets_dict": {
        "random": [1, 100]
    },
    "first_cycle_keep_size": 10,
    "num_runs": 50,
    "max_final_dets": 250,
    "max_rounds": 2,         
    "pool_core_ratio": 10,   
    "core_set_ratio": 1.1
}

final_energy, final_dets, final_coeffs, details, cfg = run_full_calculation(
    molecule=MOLECULE,
    basis=BASIS,
    spin=SPIN,
    #trimci_config_path=str(config_path),  # Optional: use config file
    config_dict=config_dict, # Optional: use config dict
)

print("=== TrimCI Result ===")
print(f"Molecule: {MOLECULE}, basis: {BASIS}")
print(f"Final energy (Ha): {final_energy:.8f}")
print(f"Final determinants: {len(final_dets)}")
print(f"Results dir: {details.get('results_dir', 'results')}")
```

#### 8) We did it! We found -2064.42292204 with 257 determinants in the final state, mathcing the results from the Hartree-Fock state. More details can be seen in the multi_run_report in the same folder.

#### 9) We may load the result and refine it.

```python
from pathlib import Path
results_dir = Path(details.get('results_dir', 'results'))
initial_dets_path = str(results_dir / "dets.npz")
print(initial_dets_path)
```

```python
from trimci import run_full_calculation

config_dict = {
    "load_initial_dets": True,
    "initial_dets_path": initial_dets_path,
    "first_cycle_keep_size": 10,
    "num_runs": 1,
    "max_final_dets": 250,
    "max_rounds": 2,         
    "pool_core_ratio": 10,   
    "core_set_ratio": 1.05
}

final_energy, final_dets, final_coeffs, details, cfg = run_full_calculation(
    molecule=MOLECULE,
    basis=BASIS,
    spin=SPIN,
    #trimci_config_path=str(config_path),  # Optional: use config file
    config_dict=config_dict, # Optional: use config dict
)

print("=== TrimCI Result ===")
print(f"Molecule: {MOLECULE}, basis: {BASIS}")
print(f"Final energy (Ha): {final_energy:.8f}")
print(f"Final determinants: {len(final_dets)}")
print(f"Results dir: {details.get('results_dir', 'results')}")
```

#### 10) It gives a better result! -2064.44228953

#### 11) Let's look at a harder problem. (Fe4S4)
